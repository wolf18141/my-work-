```python
def is_even(n) :
    return(n % 2 == 0)
print(is_even(4))
print(is_even(7))

def square(n) :
    return (n*n )
print(square(7))

def average(values) :
    if len (values) == 0:
       return None
    return (sum(values) / len(values))
print(round(average([82,90,76]),2))
```

```python
def greet(name, greeting="Ahlan"):
    return greeting , name
print(greet("sara"))
print(greet("Omar","Marhaba"))
print(greet(name = "ali",greeting ="hi"))
```

```python
def min_max(values):
   highest = values[0]
   lowest = values[0]
   for v in values:
      if v > highest :
         highest = v
      if v < lowest :
         lowest = v
   return lowest, highest
low , high = min_max([82, 90, 76,])
print(low , high)
```
