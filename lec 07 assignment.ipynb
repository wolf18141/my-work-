```python
import numpy as np

x = np.array([12.0, 18.0, 9.0, 21.0, 15.0])
norm = (x - x.mean()) / x.std()
print(round(norm.mean(), 6))
print(round(norm.std(), 6))
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m.shape)
print(m.mean(axis=0))
print(m.mean(axis=1))
```

```python
assert abs(norm.mean()) < 1e-9, "المتوسط ~ صفر"
assert abs(norm.std() - 1.0) < 1e-9, "الانحراف = 1"
assert m.sum() == 21 and m.shape == (2, 3)
print("A1: ممتاز")
```

```python
import os
import pandas as pd

os.makedirs("data", exist_ok=True)
coffee = [
    {"day": "Mon", "drink": "espresso", "cups": 12, "revenue": 36.0},
    {"day": "Mon", "drink": "latte",    "cups": 18, "revenue": 90.0},
    {"day": "Tue", "drink": "espresso", "cups": 10, "revenue": 30.0},
    {"day": "Tue", "drink": "latte",    "cups": 22, "revenue": 110.0},
    {"day": "Wed", "drink": "espresso", "cups": 14, "revenue": 42.0},
    {"day": "Wed", "drink": "tea",      "cups": 9,  "revenue": 27.0},
    {"day": "Thu", "drink": "latte",    "cups": 20, "revenue": 100.0},
    {"day": "Fri", "drink": "espresso", "cups": 16, "revenue": 48.0},
    {"day": "Fri", "drink": "latte",    "cups": 25, "revenue": 125.0},
    {"day": "Sat", "drink": "latte",    "cups": 30, "revenue": 150.0},
    {"day": "Sat", "drink": "tea",      "cups": 12, "revenue": 36.0},
]
pd.DataFrame(coffee).to_csv("data/coffee_demo.csv", index=False)

df = pd.read_csv("data/coffee_demo.csv")
print(df.shape)                                  # (11, 4)
print(df["drink"].nunique())                     # 3
print(round(df["cups"].mean(), 1))               # 17.1
print(df["revenue"].sum())                       # 794.0
best_day = df.groupby("day")["revenue"].sum().idxmax()
print(best_day)                                  # Sat
```

```python
assert df.shape == (11, 4), "11 صفوف 4 أعمدة"
assert df["drink"].nunique() == 3, "3 مشروبات"
assert round(df["cups"].mean(), 1) == 17.1, "متوسط الأكواب"
assert df["revenue"].sum() == 794.0, "إجمالي الإيراد"
assert best_day == "Sat", "أعلى يوم"
print("A2: ممتاز")
```

```python
fri = df[df["day"] == "Fri"].sort_values("revenue", ascending=False)
print(fri["drink"].iloc[0])                      # latte
assert list(fri["revenue"]) == [125.0, 48.0], "إيراد الجمعة"
top = df.sort_values("revenue", ascending=False).head(1)
assert top["day"].iloc[0] == "Sat", "أعلى صف السبت"
print("A3: ممتاز")
```

```python
by_day = df.groupby("day")["revenue"].sum().sort_values(ascending=False)
by_drink = df.groupby("drink")["cups"].mean()
print(by_day.idxmax(), by_day.max())
print(round(by_drink["latte"], 1))
```

```python
assert by_day.idxmax() == "Sat" and by_day.max() == 186.0
assert round(by_drink["latte"], 1) == 23.0, "اللاتيه"
assert round(by_drink["espresso"], 1) == 13.0, "الإسبريسو"
print("A4: ممتاز")
```

```python
messy = [
    {"date": "2026-01-01", "category": "food",  "amount": 85.5},
    {"date": "2026-01-02", "category": None,    "amount": 40.0},
    {"date": "2026-01-02", "category": None,    "amount": 40.0},
    {"date": "2026-01-03", "category": "drink", "amount": "n/a"},
    {"date": "2026-01-04", "category": "food",  "amount": 120.0},
]
pd.DataFrame(messy).to_csv("data/messy_demo.csv", index=False)

raw = pd.read_csv("data/messy_demo.csv")
print(raw.isna().sum().sum())                    # 3
dedup = raw.drop_duplicates()
print(len(dedup))                                # 4
clean = dedup.dropna()
print(len(clean))                                # 2
print(clean["amount"].astype(float).sum())       # 205.5
```

```python
assert raw.isna().sum().sum() == 3, "3 قيم ناقصة"
assert len(dedup) == 4, "دبل واحد اتمس"
assert len(clean) == 2, "صفين نضاف"
assert clean.isna().sum().sum() == 0, "مفيش نواقص"
print("A5: ممتاز")
```
