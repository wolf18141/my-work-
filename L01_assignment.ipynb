### المشكلة 1 — تحويل الوقت (أرقام: `//` و `%`)
</div>

<div dir="rtl">
لديك 205 دقيقة من المذاكرة. حوّلها إلى ساعات ودقائق: القسمة الصحيحة تكتب ساعات، والباقي يكتب دقائق. أكمل السطرين المعلّمين ثم اطبع الجملة.
</div>

<div dir="rtl">
**تذكير سريع:** `//` القسمة الصحيحة تُسقط الكسر (مثل `10 // 3` = 3)، و `%` باقي القسمة (مثل `10 % 3` = 1).
</div>

````python
total_minutes = 205
hours = total_minutes//60      # أكمل: القسمة الصحيحة
rest = total_minutes % 60        # أكمل: الباقي
print(f"{total_minutes} minutes = {hours} hours {rest} minutes")
````

**المخرجات المتوقعة:**

```text
205 minutes = 3 hours 25 minutes
```

````python
# تحقق ذاتي
assert hours == 3
assert rest == 25
````

<div dir="rtl">
### المشكلة 2 — متوسط درجات الطقس (أعداد عشرية + `round`)
</div>

<div dir="rtl">
درجات ثلاثة أيام: 21.5 و 23.0 و 19.8. احسب المتوسط، ثم قرّبه لرقم عشري واحد بـ `round`، ثم اطبع الجملة.
</div>

<div dir="rtl">
**تذكير سريع:** المتوسط = مجموع القيم ÷ عددها، و `round(x, 1)` تقرّب لرقم عشري واحد (جرّبناها مع `0.1 + 0.2` في الحصة).
</div>

````python
t1 = 21.5
t2 = 23.0
t3 = 19.8
average = (t1+t2+t3)/3      # أكمل: المتوسط ثم round لرقم عشري واحد
average = round(average,1)
print(f"Average temperature: {average} °C")
````

**المخرجات المتوقعة:**

```text
Average temperature: 21.4 °C
```

````python
# تحقق ذاتي
assert average == 21.4
````

### المشكلة 3 — تنظيف اسم دولة (نصوص)
</div>

<div dir="rtl">
اسم دولة بفراغات زائدة وأحرف صغيرة: نظّفه (`strip()` ثم `title()`)، ثم استخرج أول ثلاثة حروف، ثم احسب عدد أحرفه بـ `len()`. اطبع الثلاثة مفصولة بـ ` / `.
</div>

<div dir="rtl">
**تذكير سريع:** `.strip()` يزيل الفراغات من الطرفين، و `.title()` يجعل أول حرف كل كلمة كبيرًا، و `cleaned[0:3]` تأخذ أول ثلاثة أحرف، و `len()` تُرجع عدد الأحرف.
</div>

````python
country = "  new zealand  "
cleaned = country.strip().title()     # أكمل: strip().title()
first3 =  cleaned[0:3]     # أكمل: أول ثلاثة حروف
letters = len(cleaned)      # أكمل: len(cleaned)
print(f"{cleaned} / {first3} / {letters}")
````

**المخرجات المتوقعة:**

```text
New Zealand / New / 11
```

````python
# تحقق ذاتي
assert cleaned == "New Zealand"
assert first3 == "New"
assert letters == 11
````

### المشكلة 4 — فاتورة باقة إنترنت (تحويل + f-string)
</div>

<div dir="rtl">
نصّان يمثّلان الأرقام: ساعات الاستخدام `"2"` وسعر الساعة `"75"`. حوّلهما إلى أرقام، احسب الإجمالي، ثم اطبع الجملة بـ f-string.
</div>

<div dir="rtl">
**تذكير سريع:** النص لا يُحسب معه رقم أصلًا — حوّله بـ `int("75")` قبل الحساب (لأن `input()` ترجع نصًا دائمًا)، ودمج نص مع رقم بدون تحويل يعطي `TypeError`.
</div>

````python
from typing_extensions import reveal_type
hours_text = int("2")
rate_text = int("75")
hours = hours_text        # أكمل: int(hours_text)
rate = rate_text         # أكمل: int(rate_text)
total = hours_text * rate_text        # أكمل: hours * rate
print(f"You used {hours} hours at {rate} EGP/hour = {total} EGP")
hours_text = 2
rate_text = 75
hours = hours_text
rate = rate_text
total = hours_text * rate_text
print(f"you used {hours} hours at{rate} EGP/hour = {total} EGP" )
````

**المخرجات المتوقعة:**

```text
You used 2 hours at 75 EGP/hour = 150 EGP
```

````python
# تحقق ذاتي
assert total == 150
assert hours == 2
assert rate == 75
````

### المشكلة 5 — تصحيح 3 مقاطع (قراءة الأخطاء)
</div>

<div dir="rtl">
لكل مقطع: رسالة الخطأ الفعلية وسببها في التعليق. اكتب الإصلاح في الخلية المخصّصة، ثم شغّل خلية التحقق.
</div>

<div dir="rtl">
**تذكير سريع:** اقرأ الرسالة من آخرها — `NameError` اسم غير معرّف، `TypeError` نوع لا يصلح للعملية، و `SyntaxError` البايثون لم يفهم الجملة فلا يسمح بالتشغيل.
</div>

**المخرجات المتوقعة بعد الإصلاح:**

```text
The trip lasts 4 days
Seat 7
Cairo
```

````python
# 5.1 — NameError: name 'days' is not defined  (المقطع الأصلي: print(days) قبل تعريفه)
# الرحلة 4 أيام — عرّف المتغيّر ثم اطبع الجملة
days = 4         # أكمل
print(f"The trip lasts {days} days")
````

````python
# تحقق ذاتي
assert days == 4
````

````python
# 5.2 — TypeError: can only concatenate str (not "int") to str
# المقطع المكسور: fixed_text = "Seat " + seat_no
seat_no = 7
fixed_text = "seat" +" " + str(seat_no)  # أكمل الإصلاح
print(fixed_text)
seat_no = "7"
fixed_text = "seat" + " " + seat_no
print(fixed_text)
````

````python
# تحقق ذاتي
assert fixed_text == "Seat 7"
````

````python
# 5.3 — SyntaxError: unterminated string literal
# المقطع المكسور: destination = "Cairo
destination = "cairo" # أكمل: اكتب الوجهة المطلوبة
print(destination)
````

````python
# تحقق ذاتي
assert destination == "Cairo"
````

## قطعة المشروع (Project Piece) — تُبنى بعد الحصة
</div>

<div dir="rtl">
عرّف مشروعك كمتغيّرات ونصوص، ثم اطبع بطاقة المشروع كاملة. <code>project_name</code> و <code>topic</code> غيّرهما بموضوعك. **5 إلى 8 حقول** منفصلة كنصوص (القوائم في المحاضرة 2)، منها على الأقل **حقلان رقميان وحقل تصنيف واحد** — القيم أدناه نموذج جاهز (المقهى) لترى الشكل، استبدلها بقيم مشروعك.
</div>

````python
project_name =" data structure"          # غيّره إلى اسم مشروعك
topic = "company assignment"             # غيّره إلى موضوعك

field1_name = "order_id"
field1_sample = 101
field2_name = "item"
field2_sample = "Latte"
field3_name = "price"
field3_sample = 4.5
field4_name = "quantity"
field4_sample = 3
field5_name = "payment"
field5_sample = "card"
field6_name = "day"
field6_sample = "Monday"

# اطبع بطاقة المشروع: الاسم + الموضوع + كل حقل مع قيمته ونوعه
print("=" * 44)
print("PROJECT CARD:", project_name)
print("Topic:", topic)
print("-" * 44)
print(f"{field1_name} | sample: {field1_sample} | type: {type(field1_sample).__name__}")
print(f"{field2_name} | sample: {field2_sample} | type: {type(field2_sample).__name__}")
print(f"{field3_name} | sample: {field3_sample} | type: {type(field3_sample).__name__}")
print(f"{field4_name} | sample: {field4_sample} | type: {type(field4_sample).__name__}")
print(f"{field5_name} | sample: {field5_sample} | type: {type(field5_sample).__name__}")
print(f"{field6_name} | sample: {field6_sample} | type: {type(field6_sample).__name__}")
print("=" * 44)
````

````python
# تحقق من بنية البطاقة
assert project_name != "" and topic != ""
assert field1_name != "" and field2_name != "" and field3_name != ""
assert field4_name != "" and field5_name != "" and field6_name != ""
assert type(field1_sample).__name__ in ("int", "float", "str")
assert type(field3_sample).__name__ in ("int", "float", "str")
````
