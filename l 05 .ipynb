```python
class student :
     def __init__(self, name, grades):
         self.name = name
         self.grades = grades
         return None
     def average (self) :
        if not (self.grades) :
           return 0
        return round(sum(self.grades) / len(self.grades) , 2)
     def __str__(self):
      return f"{self.name}-{self.grades}-{self.average():.2f}"

s = student("sara", [85, 90, 76])
o = student("omar", [70, 80 ])
print(s.name,s.average(),"\\" ,o.name, o.average())

class classroom :
  def __init__(self , room) :
    self.room = room
    self.students = []
    return None
  def add(self,student) :
    self.students.append(student)
    return None
  def best(self) :
    if not self.students :
      return None
    best_student = self.students[0]
    for student in self.students :
      if student.average() > best_student.average():
        best_student = student
    return best_student

room1 = classroom("A101")
room1.add(s)
room1.add(o)
print("Best student:", room1.best())
```
